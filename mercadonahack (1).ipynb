```python


from itertools import combinations
import numpy as np

# ==== FUNCIONES BASE ====

def total_macros(menu):
    macros = {"calories": 0, "protein": 0, "carbs": 0, "fat": 0}
    for meal in menu:
        for k in macros:
            macros[k] += meal["macros"][k]
    return macros

def score_macros(target, actual):
    return -sum(abs(target[k] - actual[k]) for k in target)

def plan_day(user, recipes):
    candidates = recipes
    best_menu = None
    best_score = -np.inf

    # Convertir user (tuple) a un diccionario de macros
    target_macros = {"calories": user[0], **user[1]}

    for combo in combinations(candidates, 3):
        total = total_macros(combo)
        s = score_macros(target_macros, total)
        if s > best_score:
            best_score = s
            best_menu = combo
    return best_menu

def plan_week(user, recipes):
    week_menu = {}
    used_recipes = set()
    
    dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]

    for day in dias:
        # Filtrar recetas no repetidas
        available_recipes = [r for r in recipes if r['name'] not in used_recipes]
        
        if len(available_recipes) < 3:
            # Reiniciar si no hay suficientes recetas no repetidas
            used_recipes = set()
            available_recipes = recipes[:]
        
        daily_menu = plan_day(user, available_recipes)
        
        if daily_menu:
            week_menu[day] = daily_menu
            for r in daily_menu:
                used_recipes.add(r['name'])
        else:
            week_menu[day] = ["No hay suficientes recetas para planificar este día"]
    
    return week_menu

# ==== EJEMPLO DE USO ====

# Diccionario de usuarios


# Lista de recetas de ejemplo
recipes = [
    {'name': 'Pollo con arroz', 'macros': {'calories': 700, 'protein': 60, 'carbs': 80, 'fat': 15}},
    {'name': 'Ensalada de atún', 'macros': {'calories': 500, 'protein': 45, 'carbs': 20, 'fat': 25}},
    {'name': 'Tortilla', 'macros': {'calories': 600, 'protein': 35, 'carbs': 10, 'fat': 40}},
    {'name': 'Pasta boloñesa', 'macros': {'calories': 800, 'protein': 50, 'carbs': 90, 'fat': 30}},
    {'name': 'Salmón con verduras', 'macros': {'calories': 650, 'protein': 55, 'carbs': 25, 'fat': 35}},
    {'name': 'Batido de proteínas', 'macros': {'calories': 300, 'protein': 30, 'carbs': 15, 'fat': 5}},
    {'name': 'Tacos de ternera', 'macros': {'calories': 750, 'protein': 40, 'carbs': 60, 'fat': 40}},
    {'name': 'Pizza casera', 'macros': {'calories': 850, 'protein': 45, 'carbs': 90, 'fat': 35}},
    {'name': 'Sopa de lentejas', 'macros': {'calories': 550, 'protein': 25, 'carbs': 50, 'fat': 20}},
    {'name': 'Huevos revueltos', 'macros': {'calories': 400, 'protein': 20, 'carbs': 5, 'fat': 30}}
]

# Planificar semana para Pablo
menu_semana_pablo = plan_week(ind['Pablo'], recipes)

# Mostrar resultados
for dia, comidas in menu_semana_pablo.items():
    print(f"\n{dia}:")
    if isinstance(comidas, list) and isinstance(comidas[0], str):
        print(comidas[0])
    else:
        for comida in comidas:
            print(f" - {comida['name']}")
```
